## Create a function which return the list of gene + distance from target around a target gene (yabby)

```python
def around_gene_intra(bed, gene, gap):
    index_gene = bed[bed.gene_id == gene].index[0]
    around_gene = list(bed.loc[bed.index.isin(range(index_gene-gap,index_gene+gap))].names)
    around_gene_dist = [abs(i) for i in (list(range(-gap,gap)))]
    return([around_gene, around_gene_dist])


gap = 50

for specie in species:
    bed_table = bed_dict[specie]
    for gene_name in gene_dict[specie] :
        around_gene_dict[specie][gene_name] = around_gene_intra(bed_table, gene_name, gap)
```

```python

##function which return the blast table between two species. 

def return_blast(match):
    if match[0] == match[1]:
        return(blast_dict["blast_"+match[0]+"_"+match[1]]) #si les deux même espèces
    else:
        for i in blast_list_name: #combinaison des deux espèces
            if all([x in i for x in match]):
                return(blast_dict[i])

#function which return the list of the corresponding genes. = genes which blast together.
# def blast_list_func(blast, around_gene1, around_gene2):
#     res=[]
#     dist=[]
#     for gene in around_gene1[0]:
#         for query in blast[blast.db == gene]["query"].values:
#             if query in around_gene2[0]:
#                 res.append([gene, query])
#                 dist.append([around_gene1[1][around_gene1[0].index(gene)], around_gene2[1][around_gene2[0].index(query)]])

#     for gene in around_gene1[0]:
#         for query in blast[blast["query"] == gene]["db"].values:
#             if query in around_gene2[0]:
#                 res.append([gene, query])
#                 dist.append([around_gene1[1][around_gene1[0].index(gene)], around_gene2[1][around_gene2[0].index(query)]])
#     return([res, dist])



###chatgpt improved
def blast_list_func(blast, around_gene1, around_gene2):
    res = []
    dist = []
    gene1_set = set(around_gene1[0])
    gene2_set = set(around_gene2[0])

    for gene in gene1_set:
        if gene in blast.db.values:
            queries = blast[blast.db == gene]["query"].values
            common_queries = set(queries) & gene2_set

            for query in common_queries:
                res.append([gene, query])
                dist.append([around_gene1[1][around_gene1[0].index(gene)], around_gene2[1][around_gene2[0].index(query)]])

    for gene in gene1_set:
        if gene in blast["query"].values:
            databases = blast[blast["query"] == gene]["db"].values
            common_databases = set(databases) & gene2_set

            for query in common_databases:
                res.append([gene, query])
                dist.append([around_gene1[1][around_gene1[0].index(gene)], around_gene2[1][around_gene2[0].index(query)]])

    return [res, dist]

# def blast_list_func_same_specie(blast, around_gene1, around_gene2):
#     res=[]
#     dist=[]
#     for gene in around_gene1[0]:
#         for query in blast[blast.db == gene]["query"].values:
#             if query in around_gene2[0]:
#                 res.append([gene, query])
#                 dist.append([around_gene1[1][around_gene1[0].index(gene)], around_gene2[1][around_gene2[0].index(query)]])
#     return([res, dist])


def blast_list_func_same_specie(blast, around_gene1, around_gene2):
    res = []
    dist = []
    gene1_set = set(around_gene1[0])
    gene2_set = set(around_gene2[0])

    for gene in gene1_set:
        if gene in blast.db.values:
            queries = blast[blast.db == gene]["query"].values
            common_queries = set(queries).intersection(gene2_set)

            for query in common_queries:
                res.append([gene, query])
                dist.append([around_gene1[1][around_gene1[0].index(gene)], around_gene2[1][around_gene2[0].index(query)]])

    return [res, dist]

```

```python
nb_anchors = 3
dist_close_anchors = gap

###chatgpt improved

results_species_genes = []
results_species_genes_close = []
results_species_genes_close_target = []



for i in range(len(species)):  # boucle sur les espèces
    specie_a = species[i]
    for g1 in range(len(gene_dict[specie_a])):  # boucle sur les gènes de l'espèce 1
        gene1 = gene_dict[specie_a][g1]
        for j in range(i, len(species)): #boucle pour la deuxième espèce qui commence de l'espèce en cours pour ne pas répéter
            specie_b = species[j] 
            blast_table = return_blast([species_name_dict[specie_a], species_name_dict[specie_b]])  #on prend le blast correspondant à espèce 1 et espèce 2
            for g2 in range(len(gene_dict[specie_b])): #on boucle sur les gènes de l'espèce 2
                gene2 = gene_dict[specie_b][g2]
                if specie_a == specie_b: #si la même espèce on ne répète pas les mêmes combinaisons.
                    if g1 >= g2:
                        continue
                    else:
                        result = blast_list_func_same_specie(blast_table, around_gene_dict[specie_a][gene1], around_gene_dict[specie_b][gene2])
                else:
                    result = blast_list_func(blast_table, around_gene_dict[specie_a][gene1], around_gene_dict[specie_b][gene2])
                anchors, dist = result[0], result[1]
                if len(anchors) >= nb_anchors:  # si au moins 3 ancrages
                    results_species_genes.append([specie_a, specie_b, gene1, gene2, result])
                    dist_g1 = len([i[0] for i in dist if i[0] < dist_close_anchors])
                    dist_g2 = len([i[1] for i in dist if i[1] < dist_close_anchors])
                    if dist_g1 > 2 and dist_g2 > 2:
                        results_species_genes_close.append([specie_a, specie_b, dict_gene_yabby[gene1], dict_gene_yabby[gene2], dist_g1])
                        if gene1 in list(itertools.chain.from_iterable(anchors)) and gene2 in list(itertools.chain.from_iterable(anchors)):
                            results_species_genes_close_target.append([specie_a, specie_b, dict_gene_yabby[gene1], dict_gene_yabby[gene2], dist_g1])

```
