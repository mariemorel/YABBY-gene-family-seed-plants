```python
prot_file =path+"Infer_Anth/data/gymnosperms/complete_genomes/pine/P.tabuliformis_V1.0.Protein.fa"
bed_file = path+"Infer_Anth/data/gymnosperms/complete_genomes/pine/P.tabuliformis_V1.0.mRNA.bed"
merge_bed_file = path+"Infer_Anth/data/gymnosperms/complete_genomes/pine/P.tabuliformis_V1.0.mRNA.subset.sort.merge.bed"
```

```python
merge_BED = pd.read_csv(merge_bed_file, sep="\t", header=None, names=["chr", "start", "end", "id", "strand"], on_bad_lines="skip" )
merge_BED.head()
```

```python
BED = pd.read_csv(bed_file, sep="\t", header=None, names=["chr", "start", "end", "id","unk1" , "strand"], on_bad_lines="skip" )
BED.head()
```

```python
unique_seq = pd.DataFrame(merge_BED[[len(i.split(",")) == 1 for i in merge_BED.id.values]])
spliced_seq = pd.DataFrame(merge_BED[[len(i.split(",")) > 1 for i in merge_BED.id.values]])
```

```python
unique_seq.head()
```

```python
unique_seq["names"]= [i.split(";")[0].split("=")[1] for i in unique_seq.id]
unique_seq.head()
```

```python
#BED.to_csv(path+"Infer_Anth/data/gymnosperms/complete_genomes/pine/P.tabuliformis_V1.0.mRNA.subset.bed", sep="\t", header=None, index=None) ##no bad lines
```

```python
records = SeqIO.parse(prot_file, "fasta")
sequences = []
for rec in records:
    seq = SeqRecord(rec.seq, id=rec.name ,name="",description="")
    sequences.append(seq)


fasta_names = [seq.id for seq in sequences]
sequences_dict = {i.id:i.seq for i in sequences}
```

```python
spliced_seq.head()
```

```python
##take name with longest sequence
uniq_name = []
not_in_fasta = []
for seq in spliced_seq["id"]:
    try :   
        A_list = [len(sequences_dict[name.split(";")[0].split("=")[1]]) for name in seq.split(",")]
        A_index = A_list.index(max(A_list)) 
        uniq_name.append(seq.split(",")[A_index])
    except KeyError:
        not_in_fasta.append(seq)
```

```python
uniq_name[:10]
```

```python
uniq_name_bis = []
for splice in not_in_fasta: 
    in_fasta = [seq.split(";")[0].split("=")[1] for seq in splice.split(",") if seq.split(";")[0].split("=")[1] in sequences_dict]
    if len(in_fasta) > 0:
        A_list = [len(sequences_dict[name]) for name in in_fasta]
        A_index = A_list.index(max(A_list)) 
        uniq_name_bis.append(splice.split(",")[A_index])
```

```python
uniq_name_bis
```

```python
all_spliced = uniq_name+uniq_name_bis
```

```python
len(all_spliced)
```

```python
BED_unique = pd.DataFrame(BED[BED.id.isin(all_spliced)])
```

```python
BED_unique["names"]= [i.split(";")[0].split("=")[1] for i in BED_unique.id]
BED_unique.drop(columns=['unk1'], inplace = True)
BED_unique.head()
```

```python
BED_nosplicing = pd.concat([unique_seq, BED_unique], ignore_index=True)
```

```python
BED_nosplicing.head()
```

```python
longest_sequences = dict()
for id in sequences_dict:
    if id in BED_nosplicing.names.values:
        longest_sequences[id] = sequences_dict[id]
```

```python
sequences = []
for rec in longest_sequences:
    seq = SeqRecord(longest_sequences[rec], id=rec ,name="",description="")
    sequences.append(seq)

fasta_names = [seq.id for seq in sequences]
```

```python
with open(path+"Infer_Anth/data/gymnosperms/complete_genomes/pine/pinus_nosplicing.pep.fa", "w") as output_handle:
    SeqIO.write(sequences, output_handle, "fasta")
```

```python
BED_nosplicing_fasta = pd.DataFrame(BED_nosplicing[BED_nosplicing.names.isin(fasta_names)])
```
